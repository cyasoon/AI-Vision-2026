```python
def calculate_bmi():
    # Prompt the user for weight in kilograms and height in meters
    try:
        weight = float(input("Enter your weight in kg: "))
        height = float(input("Enter your height in meters: "))

        if weight <= 0 or height <= 0:
            print("Weight and height must be positive numbers.")
            return

        # BMI Formula: weight / (height^2)
        bmi = weight / (height ** 2)

        # Classify the BMI result
        if bmi < 18.5:
            category = "Underweight"
        elif 18.5 <= bmi < 25.0:
            category = "Normal"
        elif 25.0 <= bmi < 30.0:
            category = "Overweight"
        else:
            category = "Obese"

        # Output the results rounded to 2 decimal places
        print(f"\nYour BMI is: {bmi:.2f}")
        print(f"Classification: {category}")

    except ValueError:
        print("Invalid input. Please enter numerical values for weight and height.")

# Run the program
if __name__ == "__main__":
    calculate_bmi()
```
