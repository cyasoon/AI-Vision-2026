### c. gunakan data markah tadi untuk menghasilkan bar chart / graf perlu menunjukkan nama pelajar di paksi X dan Markah di Paksi Y:

```python
import numpy as np

# Data asal
markah_array = np.array([80, 75, 90, 85, 70])

print("Semua markah:")
print(markah_array)
print()

print("Markah tertinggi:")
print(max(markah_array))
print()

print("Markah terendah:")
print(min(markah_array))
print()

print("Purata markah:")
print(np.mean(markah_array))
print()

print("Jumlah (sum):")
print(sum(markah_array))
```

```python
import numpy as np
import pandas as pd

# data nama dan markah
markah_array = np.array([80, 75, 90, 85, 70])
senarai_nama = ["Pelajar A", "Pelajar B", "Pelajar C", "Pelajar D", "Pelajar E"]

# gabungkan untuk jadi dataframe
data = {"Nama Pelajar": senarai_nama, "Markah": markah_array}
df = pd.DataFrame(data)

print("Jadual Markah Pelajar:")
print()
print(df)
```

```python
import matplotlib.pyplot as plt

# data untuk paksi x dan paksi y
senarai_nama = ["Pelajar A", "Pelajar B", "Pelajar C", "Pelajar D", "Pelajar E"]
markah_array = [80, 75, 90, 85, 70]

# lukis graf bar
plt.bar(senarai_nama, markah_array, color=["red", "blue", "green", "orange", "purple"])

plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.title("Graf Prestasi Markah Pelajar")

# papar graf
plt.show()
```
