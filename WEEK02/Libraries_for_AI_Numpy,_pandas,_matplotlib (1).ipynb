#`matplotlib` untuk plot graf

```python
import matplotlib.pyplot as plt

marks = [80,90,75]

plt.plot(marks)

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.plot(nama, markah, marker='o')

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.bar(nama, markah, color=["blue", "pink", "green"])

plt.title("Markah Ujian Pelajar")

plt.xlabel("Nama Pelajar")
plt.ylabel("Markah (%)")

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.scatter(nama, markah)

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.bar(nama, markah, label="Sesi 1")

plt.bar(nama, markah, bottom=markah, label="Sesi 2")

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.hist(markah)

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.pie(markah, labels=nama)

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.pie(markah, labels=nama, autopct='%.1f%%')

plt.title("Pecahan Markah Pelajar (Peratus)")

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.stackplot(nama, markah, colors=['lightblue'])

# Tambah label sedikit
plt.title("Graf Kawasan Markah")
plt.xlabel("Nama")
plt.ylabel("Markah")

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.stem(nama, markah)

plt.title("Graf Batang-dan-Daun Markah")
plt.xlabel("Nama")
plt.ylabel("Markah")

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.barh(nama, markah)

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar"]
markah = [80, 90, 75]

plt.step(nama, markah)

plt.show()
```
